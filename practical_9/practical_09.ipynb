```python
initial_state = {"At_A"}

goal = {"At_C"}
```

```python
actions = [
  {
    "name": "Move A to B",
    "pre": {"At_A"},
    "add": {"At_B"},
    "delete": {"At_A"}
  },
  {
    "name": "Move B to C",
    "pre": {"At_B"},
    "add": {"At_C"},
    "delete": {"At_B"}
  }
]
```

```python
def find_actions(goal):
  for action in actions:
    if goal in action["add"]:
      return action
  return None
```

```python
def make_plan(goal,state,plan):
  if goal in state:
    return True

  action = find_actions(goal)

  if action is None:
    return False

  for condition in action["pre"]:
    if not make_plan(condition, state, plan):
      return False

  state.difference_update(action["delete"])
  state.update(action["add"])
  plan.append(action["name"])
  return True
```

```python
# To fix the issue, we extract the string "At_C" from your goal set
single_goal = list(goal)[0]
state = initial_state.copy()
plan = []

# Run the planner with the string target "At_C"
make_plan(single_goal, state, plan)

print("Plan:")
for step in plan:
    print(step)
```
